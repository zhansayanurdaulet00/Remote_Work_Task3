```python
import pandas as pd
import numpy as np

# Excel файлын оқу
df = pd.read_excel("Новая форма (Ответы).xlsx")

# Уақыт белгісінен басқа барлық бағандар бос жолдарды алып тастау
response_columns = df.columns.drop("Отметка времени")

df_clean = df.dropna(
    subset=response_columns,
    how="all"
).copy()

print("Бастапқы жол саны:", len(df))
print("Тазартудан кейінгі жол саны:", len(df_clean))
print("Қайталанатын жолдар:", df_clean.duplicated().sum())

display(df_clean.head())
```

```python

import pandas as pd
import numpy as np
import re

# Тазартылған деректердің көшірмесі
df_clean = df_clean.copy()

# Бағандарды қысқа атауларға ауыстыру
cols = list(df_clean.columns)

rename = {
    cols[1]: "age",
    cols[2]: "work_arrangement",
    cols[3]: "remote_days",
    cols[4]: "remote_percentage",
    cols[5]: "weekly_hours",
    cols[6]: "remote_experience"
}

df_clean = df_clean.rename(columns=rename)

# Q1: Жас
df_clean["age"] = pd.to_numeric(
    df_clean["age"], errors="coerce"
)

# Q2: Жұмыс форматы
df_clean["work_arrangement"] = (
    df_clean["work_arrangement"].astype("string").str.strip()
)

# Q3: Қашықтан жұмыс күндері
def clean_remote_days(x):
    if pd.isna(x):
        return np.nan
    s = str(x).lower().strip()

    if s in ["ешқашан", "жоқ",
             "қашықтықтан жасамаймын",
             "қашықтықтан жұмыс істемеймін"]:
        return 0

    if "әртүр" in s:
        return np.nan

    nums = re.findall(r"\d+", s)
    if not nums:
        return np.nan

    values = [int(n) for n in nums]
    return sum(values) / len(values)

df_clean["remote_days"] = df_clean["remote_days"].apply(clean_remote_days)

# Q4: Қашықтан жұмыс пайызы
df_clean["remote_percentage"] = (
    df_clean["remote_percentage"]
    .astype("string").str.replace("%", "", regex=False)
    .str.strip()
)

# Q5: Апталық жұмыс сағаты
def clean_hours(x):
    if pd.isna(x):
        return np.nan
    s = str(x).lower().strip()

    if "әртүр" in s:
        return np.nan

    nums = re.findall(r"\d+", s)
    if not nums:
        return np.nan

    values = [int(n) for n in nums]
    return sum(values) / len(values)

df_clean["weekly_hours"] = df_clean["weekly_hours"].apply(clean_hours)

# Q6: Қашықтан жұмыс тәжірибесі
def clean_experience(x):
    if pd.isna(x):
        return np.nan
    s = str(x).lower().strip()

    if s in ["жоқ", "нет", "none"]:
        return 0

    nums = re.findall(r"\d+", s)

    if "полгода" in s or "6 месяцев" in s:
        return 0.5

    if not nums:
        return np.nan

    values = [int(n) for n in nums]
    value = sum(values) / len(values)

    if "месяц" in s:
        return value / 12

    return value

df_clean["remote_experience"] = (
    df_clean["remote_experience"].apply(clean_experience)
)

print("Q1-Q6 тазартылды")
display(df_clean.iloc[:, :7].head(10))
```

```python

# Likert сұрақтары
likert_cols = list(df_clean.columns[7:31])

# Жауаптан алғашқы санды алу
def extract_likert(value):
    if pd.isna(value):
        return np.nan

    match = re.match(r"\s*([1-5])\s*[-—]", str(value))

    if match:
        return int(match.group(1))

    return np.nan

for col in likert_cols:
    df_clean[col] = df_clean[col].apply(extract_likert)

print("Likert шкаласы сандық форматқа ауыстырылды")

display(df_clean[likert_cols].head())
```

```python
# 5 негізгі зерттеу көрсеткішін есептеу

groups = {
    "boundary_blurring": ["Q7", "Q8", "Q9", "Q23", "Q24"],

    "burnout_score": ["Q10", "Q11", "Q12", "Q13", "Q25", "Q26"],

    "work_life_balance": ["Q14", "Q15", "Q16"],

    "schedule_autonomy": ["Q17", "Q18", "Q19", "Q27", "Q28"],

    "psychological_detachment": ["Q20", "Q21", "Q22", "Q29", "Q30"]
}

# Әр респондент үшін орташа балл есептеу
scores = pd.DataFrame(index=df_final.index)

for name, questions in groups.items():
    scores[name] = df_final[questions].mean(axis=1)

print("Респонденттер саны:", len(scores))
print("5 көрсеткіш есептелді")

display(scores.head(10))

print("\nDescriptive Statistics:")
display(scores.describe().round(2))
```

```python


def cronbach_alpha(data):
    data = data.dropna()
    k = data.shape[1]

    if k < 2:
        return np.nan

    item_variances = data.var(axis=0, ddof=1)
    total_variance = data.sum(axis=1).var(ddof=1)

    if total_variance == 0:
        return np.nan

    return (k / (k - 1)) * (
        1 - item_variances.sum() / total_variance
    )

print("Cronbach's Alpha — 18+ sample")

for name, questions in groups.items():
    alpha = cronbach_alpha(df_final[questions])
    print(f"{name}: {alpha:.3f}")
```

```python
# Boundary blurring сұрақтарын жеке тексеру

boundary_questions = ["Q7", "Q8", "Q9", "Q23", "Q24"]

boundary_data = df_final[boundary_questions].dropna()

print("Cronbach's Alpha:", 
      round(cronbach_alpha(boundary_data), 3))

print("\nӘр сұрақты алып тастағандағы Alpha:")

for question in boundary_questions:

    remaining_questions = [
        q for q in boundary_questions if q != question
    ]

    alpha_without = cronbach_alpha(
        boundary_data[remaining_questions]
    )

    print(question, ":", round(alpha_without, 3))
```

```python

from scipy.stats import spearmanr
from itertools import combinations
from statsmodels.stats.multitest import multipletests

results = []

variables = list(groups.keys())

for var1, var2 in combinations(variables, 2):
    rho, p_value = spearmanr(
        scores[var1],
        scores[var2],
        nan_policy="omit"
    )

    results.append({
        "Variable 1": var1,
        "Variable 2": var2,
        "Spearman rho": rho,
        "p-value": p_value
    })

correlation_results = pd.DataFrame(results)

reject, p_adjusted, _, _ = multipletests(
    correlation_results["p-value"],
    alpha=0.05,
    method="holm"
)

correlation_results["Adjusted p-value"] = p_adjusted
correlation_results["Significant after Holm"] = reject

display(
    correlation_results.round(4)
)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
from pathlib import Path

OUT = Path("domain_model_outputs")
OUT.mkdir(exist_ok=True)
DATA = Path("survey_final1 (1).xls")
if not DATA.exists():
    DATA = Path("/mnt/data/survey_final1 (1).xls")
df = pd.read_csv(DATA)
print("Rows:", len(df), "Columns:", len(df.columns))
```

```python
groups = {
    "boundary_blurring": [7,8,9,23,24],
    "burnout": [10,11,12,13,25,26],
    "work_life_balance": [14,15,16],
    "schedule_autonomy": [17,18,19,27,28],
    "psychological_detachment": [20,21,22,29,30],
}
for name, qs in groups.items():
    cols = [f"Q{i}" for i in qs]
    df[cols] = df[cols].apply(pd.to_numeric, errors="coerce")
    # Require every item answered to calculate the composite
    df[name] = df[cols].mean(axis=1).where(df[cols].notna().all(axis=1))

df["remote_days"] = pd.to_numeric(df["remote_days"], errors="coerce")
# 0–7 days/week; do not silently retain impossible values such as 10
df.loc[~df["remote_days"].between(0,7), "remote_days"] = np.nan
cols = ["burnout", "remote_days", "boundary_blurring", "work_life_balance", "schedule_autonomy", "psychological_detachment"]
model_data = df[cols].dropna().copy()
print("Model sample N:", len(model_data))
display(model_data.describe().round(3))
```

```python
predictors = ["remote_days", "boundary_blurring", "work_life_balance", "schedule_autonomy", "psychological_detachment"]
X = sm.add_constant(model_data[predictors])
y = model_data["burnout"]
model = sm.OLS(y, X).fit(cov_type="HC3")
print(model.summary())
coef = pd.DataFrame({"coefficient": model.params, "robust_se": model.bse, "p_value": model.pvalues, "ci_lower": model.conf_int()[0], "ci_upper": model.conf_int()[1]})
display(coef.round(4))
coef.to_csv(OUT / "domain_model_coefficients.csv")
model_data["predicted_burnout"] = model.predict(X)
model_data["residual"] = y - model_data["predicted_burnout"]
rmse = np.sqrt(np.mean(model_data["residual"]**2))
print(f"N={int(model.nobs)}, R²={model.rsquared:.3f}, adjusted R²={model.rsquared_adj:.3f}, in-sample RMSE={rmse:.3f}")
model_data.to_csv(OUT / "model_predictions.csv", index=False)
```

```python
fig, ax = plt.subplots(figsize=(7,6))
ax.scatter(y, model_data["predicted_burnout"], alpha=.75, edgecolor="black", linewidth=.4)
ax.plot([1,5],[1,5], linestyle="--", color="gray", label="Perfect fit")
ax.set(xlabel="Observed burnout score", ylabel="Predicted burnout score", title="Domain model: observed vs predicted burnout", xlim=(1,5), ylim=(1,5))
ax.text(.04,.95,f"N={len(model_data)}\nR²={model.rsquared:.3f}\nRMSE={rmse:.3f}",transform=ax.transAxes,va="top")
ax.legend();fig.tight_layout();fig.savefig(OUT / "observed_vs_predicted.png",dpi=200);plt.show()
```

```python
names = predictors
fig, ax = plt.subplots(figsize=(9,5))
values = model.params.loc[names]
low = model.conf_int().loc[names,0]
high = model.conf_int().loc[names,1]
ax.errorbar(values, range(len(names)), xerr=[values-low,high-values], fmt="o", capsize=4)
ax.axvline(0, color="gray", linestyle="--")
ax.set_yticks(range(len(names)), names)
ax.set(xlabel="OLS coefficient (95% HC3 CI)", title="Domain model coefficients")
fig.tight_layout();fig.savefig(OUT / "coefficient_intervals.png",dpi=200);plt.show()
```

```python

%%writefile src/process_data.py

import pandas as pd
from pathlib import Path

BASE = Path(__file__).resolve().parent.parent
INPUT = BASE / "data" / "survey_final1.csv"
OUTPUT = BASE / "data" / "analysis_ready.csv"

GROUPS = {
    "boundary_blurring": ["Q7", "Q8", "Q9", "Q23", "Q24"],
    "burnout_score": ["Q10", "Q11", "Q12", "Q13", "Q25", "Q26"],
    "work_life_balance": ["Q14", "Q15", "Q16"],
    "schedule_autonomy": ["Q17", "Q18", "Q19", "Q27", "Q28"],
    "psychological_detachment": ["Q20", "Q21", "Q22", "Q29", "Q30"]
}

REMOTE_MAPPING = {
    "0": 0,
    "1-25": 12.5,
    "26-50": 38,
    "51-75": 63,
    "76-100": 88
}


def process_data():
    df = pd.read_csv(INPUT)

    # Check duplicate responses
    print("Duplicate rows:", df.duplicated().sum())

    # Check missing values
    print("\nMissing values:")
    print(df.isna().sum())

    # Validate age
    df["age"] = pd.to_numeric(df["age"], errors="coerce")
    df = df[df["age"].ge(18)].copy()

    # Check remote working days
    df["remote_days"] = pd.to_numeric(
        df["remote_days"], errors="coerce"
    )

    df["invalid_remote_days"] = (
        df["remote_days"].notna()
        & ~df["remote_days"].between(0, 7)
    )

    # Check suspicious weekly hours
    df["weekly_hours"] = pd.to_numeric(
        df["weekly_hours"], errors="coerce"
    )

    df["hours_review"] = (
        df["weekly_hours"].lt(10)
        | df["weekly_hours"].gt(80)
        | df["weekly_hours"].isna()
    )

    # Calculate five construct scores
    for name, questions in GROUPS.items():
        df[questions] = df[questions].apply(
            pd.to_numeric, errors="coerce"
        )

        df[name] = df[questions].mean(axis=1).where(
            df[questions].notna().all(axis=1)
        )

    # Estimated remote work intensity
    df["remote_intensity"] = (
        df["remote_percentage"]
        .astype("string")
        .str.strip()
        .map(REMOTE_MAPPING)
    )

    # Save processed data
    OUTPUT.parent.mkdir(parents=True, exist_ok=True)

    df.to_csv(
        OUTPUT,
        index=False,
        encoding="utf-8-sig"
    )

    print("\nProcessing completed!")
    print("Final respondents:", len(df))
    print("Saved file:", OUTPUT)

    return df


if __name__ == "__main__":
    process_data()
```

```python

%%writefile src/visualize.py

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

BASE = Path(__file__).resolve().parent.parent
DATA = BASE / "data" / "analysis_ready.csv"
FIGURES = BASE / "figures"


def visualize():

    df = pd.read_csv(DATA)

    FIGURES.mkdir(parents=True, exist_ok=True)

    # Figure 1: Remote Work Intensity vs Burnout
    plt.figure(figsize=(8, 5))

    sns.regplot(
        data=df,
        x="remote_intensity",
        y="burnout_score",
        scatter_kws={"alpha": 0.7}
    )

    plt.title("Remote Work Intensity vs Emotional Burnout")
    plt.xlabel("Estimated Remote Work Intensity (%)")
    plt.ylabel("Emotional Burnout Score")

    plt.tight_layout()

    plt.savefig(
        FIGURES / "fig1_remote_intensity_vs_burnout.png",
        dpi=300
    )

    plt.close()

    # Figure 2: Burnout by Work Arrangement
    plt.figure(figsize=(8, 5))

    sns.boxplot(
        data=df,
        x="work_arrangement",
        y="burnout_score",
        order=["Fully on-site", "Hybrid", "Fully remote"]
    )

    plt.title("Burnout Distribution by Work Arrangement")
    plt.xlabel("Work Arrangement")
    plt.ylabel("Emotional Burnout Score")

    plt.tight_layout()

    plt.savefig(
        FIGURES / "fig2_burnout_by_arrangement.png",
        dpi=300
    )

    plt.close()

    print("Visualizations completed!")
    print("Figures saved in:", FIGURES)


if __name__ == "__main__":
    visualize()
```
